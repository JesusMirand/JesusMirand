```python
from random import random
from math import log,sqrt,pi,cos,exp
import matplotlib.pyplot as plt
def gen(n):
  l=[]
  for i in range(n):
    u1=random()
    u2=random()
    l.append(sqrt(-2*log(u1))*cos(2*pi*u2))
  return l
n=1000000
l=gen(n)
x1=-4
x2=4
x=x1
b=[]
while x<=x2:
  b.append(x)
  x=x+(x2-x1)/50
ly=[]
for x in b:
  ly.append((1/sqrt(2*pi))*exp(-x**2/2))
plt.hist(l,density=1,bins=b,color='grey',edgecolor='black')
plt.plot(b,ly,color='red')
```
